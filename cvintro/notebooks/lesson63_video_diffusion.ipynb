# Lesson 63: Video Diffusion and Generation

Lesson 62's diffusion models learn to reverse a noising process one still image at a time. This lesson extends that same denoising recipe to video: generate a whole clip at once (or frame-by-frame, autoregressively) while keeping it temporally consistent &mdash; the new failure mode a video generator has that an image generator never faces is flicker, where independently-denoised frames disagree slightly from one frame to the next. The fix is architectural: spatio-temporal attention or temporal convolution layers woven into the denoising network so nearby frames can attend to each other, plus techniques like noise scheduling shared across frames, that keep the generated motion coherent.
